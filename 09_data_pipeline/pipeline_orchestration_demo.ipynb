## Orchestration Execution Flow

The orchestration script controls the execution of the ETL pipeline.

The execution flow is:

1. The scheduler starts.
2. The scheduled job triggers the pipeline.
3. The pipeline extracts the source datasets.
4. The data is transformed and cleaned.
5. The processed datasets are loaded into Parquet files.
6. Successful and failed runs are recorded in `pipeline_run.log`.
7. If the pipeline fails, the orchestration script retries the execution up to three times.

For demonstration purposes, the pipeline is scheduled every two minutes. In a real business environment, it could run hourly, daily, or whenever new data becomes available.
