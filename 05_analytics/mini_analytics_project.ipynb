```python
analytics_data = (
    order_items
    .merge(orders, on="order_id", how="left")
    .merge(customers, on="customer_id", how="left")
    .merge(products, on="product_id", how="left")
    .merge(transactions, on="order_id", how="left")
)

print("Combined dataset shape:", analytics_data.shape)
analytics_data.head()
```

```python
print("Combined dataset shape:", analytics_data.shape)
```

```python
analytics_data.isnull().sum()
```

```python
print("ORDER STATUS")
print(orders["status"].value_counts())

print("\nPAYMENT STATUS")
print(transactions["payment_status"].value_counts())
```

```python
successful_orders = orders[orders["status"] == "completed"].copy()

successful_transactions = transactions[
    transactions["payment_status"] == "paid"
].copy()

print("Successful orders:", successful_orders.shape[0])
print("Successful transactions:", successful_transactions.shape[0])
```

```python
successful_order_ids = set(successful_orders["order_id"])
paid_order_ids = set(successful_transactions["order_id"])

fully_successful_order_ids = successful_order_ids.intersection(paid_order_ids)

print("Orders that are both completed and paid:", len(fully_successful_order_ids))
```

```python
successful_sales = analytics_data[
    analytics_data["order_id"].isin(fully_successful_order_ids)
].copy()

print("Successful sales rows:", successful_sales.shape[0])
print("Unique successful orders:", successful_sales["order_id"].nunique())
```

```python
confirmed_revenue = successful_sales["total_price"].sum()

print(f"Confirmed revenue: {confirmed_revenue:,.2f}")
```

```python
average_order_value = confirmed_revenue / len(fully_successful_order_ids)

print(f"Average order value: {average_order_value:,.2f}")
```

```python
total_items_sold = successful_sales["quantity"].sum()

print("Total items sold:", total_items_sold)
```

```python
successful_order_rate = (
    len(fully_successful_order_ids) / orders["order_id"].nunique()
) * 100

print(f"Successful order rate: {successful_order_rate:.2f}%")
```

```python
successful_customer_orders = orders[
    orders["order_id"].isin(fully_successful_order_ids)
]

orders_per_customer = (
    successful_customer_orders
    .groupby("customer_id")["order_id"]
    .nunique()
)

returning_customers = (orders_per_customer > 1).sum()
successful_customers = orders_per_customer.shape[0]

returning_customer_rate = (
    returning_customers / successful_customers
) * 100

print("Successful customers:", successful_customers)
print("Returning customers:", returning_customers)
print(f"Returning customer rate: {returning_customer_rate:.2f}%")
```

```python
payment_success_rate = (
    (transactions["payment_status"] == "paid").sum()
    / transactions["transaction_id"].nunique()
) * 100

print(f"Payment success rate: {payment_success_rate:.2f}%")
```

```python
kpi_summary = pd.DataFrame({
    "KPI": [
        "Confirmed Revenue",
        "Successful Orders",
        "Successful Order Rate",
        "Average Order Value",
        "Total Items Sold",
        "Successful Customers",
        "Returning Customer Rate",
        "Payment Success Rate"
    ],
    "Value": [
        f"{confirmed_revenue:,.2f}",
        len(fully_successful_order_ids),
        f"{successful_order_rate:.2f}%",
        f"{average_order_value:,.2f}",
        total_items_sold,
        successful_customers,
        f"{returning_customer_rate:.2f}%",
        f"{payment_success_rate:.2f}%"
    ]
})

kpi_summary
```

## Business Question 1: How does confirmed revenue change over time?

This analysis examines the daily revenue generated from orders that were both completed and successfully paid. The purpose is to identify whether revenue is stable, increasing, decreasing, or concentrated on specific days.

```python
daily_revenue = (
    successful_sales
    .groupby("order_date", as_index=False)["total_price"]
    .sum()
    .sort_values("order_date")
)

daily_revenue.head()
```

```python
successful_sales["order_day"] = successful_sales["order_date"].dt.date

daily_revenue = (
    successful_sales
    .groupby("order_day", as_index=False)["total_price"]
    .sum()
    .sort_values("order_day")
)

daily_revenue.head()
```

```python
fig = px.line(
    daily_revenue,
    x="order_day",
    y="total_price",
    markers=True,
    title="Daily Confirmed Revenue Trend"
)

fig.update_layout(
    xaxis_title="Order Date",
    yaxis_title="Confirmed Revenue"
)

fig.show()
```

```python
highest_revenue_day = daily_revenue.loc[daily_revenue["total_price"].idxmax()]
lowest_revenue_day = daily_revenue.loc[daily_revenue["total_price"].idxmin()]

print("Highest revenue day:")
print(highest_revenue_day)

print("\nLowest revenue day:")
print(lowest_revenue_day)
```

### What does this chart mean?

The daily confirmed revenue is not stable across the analysis period. Revenue rises and falls significantly instead of following a consistent upward or downward trend.

The highest confirmed revenue was **9,378.07 on 10 May 2026**, while the lowest was **203.10 on 15 April 2026**. This shows that sales performance is concentrated on certain days.

From a business perspective, the company should investigate what caused the strongest revenue days, such as larger orders, popular products, promotions, or increased customer activity. Similar strategies could then be tested on weaker days to improve revenue consistency.

## Business Question 2: Which product categories generate the most confirmed revenue?

This analysis compares confirmed revenue across product categories. The purpose is to identify which categories contribute most to successful sales and which categories may need improvement.

```python
category_revenue = (
    successful_sales
    .groupby("category", as_index=False)["total_price"]
    .sum()
    .sort_values("total_price", ascending=False)
)

category_revenue
```

```python
fig = px.bar(
    category_revenue,
    x="category",
    y="total_price",
    text_auto=".2s",
    title="Confirmed Revenue by Product Category"
)

fig.update_layout(
    xaxis_title="Product Category",
    yaxis_title="Confirmed Revenue"
)

fig.show()
```

### What does this chart mean?

The **Home** category generated the highest confirmed revenue at **25,510.71**, followed by **Clothing** at **21,629.48**.

**Books** generated the lowest confirmed revenue at **17,237.11**, although the difference between categories is not extremely large.

From a business perspective, the Home category is currently the strongest revenue contributor. The company should maintain good product availability in this category and investigate which Home products are driving the result. The weaker categories should also be reviewed to identify opportunities such as promotions, bundles, or better product positioning.

## Business Question 3: Which products generate the most confirmed revenue?

This analysis identifies the individual products contributing the most to confirmed revenue. The purpose is to determine which products are the strongest sales drivers within their categories.

```python
product_revenue = (
    successful_sales
    .groupby("product_name", as_index=False)["total_price"]
    .sum()
    .sort_values("total_price", ascending=False)
    .head(10)
)

product_revenue
```

```python
fig = px.bar(
    product_revenue.sort_values("total_price"),
    x="total_price",
    y="product_name",
    orientation="h",
    text_auto=".2s",
    title="Top 10 Products by Confirmed Revenue"
)

fig.update_layout(
    xaxis_title="Confirmed Revenue",
    yaxis_title="Product"
)

fig.show()
```

### What does this chart mean?

**Product23** generated the highest confirmed revenue at **8,570.90**, followed closely by **Product22** at **8,448.90** and **Product11** at **7,977.31**.

The chart shows that a small group of products contributes a large share of confirmed revenue. These products are important sales drivers and should receive close attention.

From a business perspective, the company should maintain enough stock for the strongest products and investigate why customers prefer them. These products could also be used in promotions, recommendations, or bundles to increase sales of related products.

## Business Question 4: Which customer cities generate the most confirmed revenue?

This analysis compares confirmed revenue across customer cities. The purpose is to identify the locations containing the highest-value customers and the strongest sales opportunities.

```python
city_revenue = (
    successful_sales
    .groupby("city", as_index=False)["total_price"]
    .sum()
    .sort_values("total_price", ascending=False)
)

city_revenue
```

```python
fig = px.bar(
    city_revenue,
    x="city",
    y="total_price",
    text_auto=".2s",
    title="Confirmed Revenue by Customer City"
)

fig.update_layout(
    xaxis_title="Customer City",
    yaxis_title="Confirmed Revenue"
)

fig.show()
```

### What does this chart mean?

**Abu Dhabi** generated the highest confirmed revenue at **26,582.41**, followed closely by **Ajman** at **26,091.77**.

**Sharjah** generated the lowest confirmed revenue at **12,577.02**, which is much lower than Abu Dhabi and Ajman.

From a business perspective, Abu Dhabi and Ajman appear to contain the strongest customer markets. The company could focus more marketing and customer-retention efforts in these cities. Sharjah should also be investigated to understand whether lower revenue is caused by fewer customers, lower order values, or weaker product demand.

## Business Question 5: What is the distribution of payment outcomes?

This analysis compares paid, failed, and refunded transactions. The purpose is to understand whether payment problems may be reducing successful sales and confirmed revenue.

```python
payment_distribution = (
    transactions["payment_status"]
    .value_counts()
    .rename_axis("payment_status")
    .reset_index(name="transaction_count")
)

payment_distribution
```

```python
fig = px.bar(
    payment_distribution,
    x="payment_status",
    y="transaction_count",
    text="transaction_count",
    title="Transaction Count by Payment Status"
)

fig.update_layout(
    xaxis_title="Payment Status",
    yaxis_title="Number of Transactions"
)

fig.show()
```

### What does this chart mean?

Only **67 transactions were successfully paid**, while **71 were refunded** and **62 failed**.

This means that most transactions did not result in a successful payment. The payment success rate is only **33.50%**, which may be limiting confirmed revenue and the number of fully successful orders.

From a business perspective, the company should investigate the main causes of failed payments and refunds. Possible areas to review include payment-method reliability, checkout problems, customer cancellations, refund reasons, and transaction-processing issues.

## Business Question 6: How many successful customers returned to purchase again?

This analysis compares one-time customers with returning customers. The purpose is to understand whether the business is successfully retaining customers after their first completed and paid order.

```python
customer_retention = pd.DataFrame({
    "customer_type": ["One-time customers", "Returning customers"],
    "customer_count": [
        successful_customers - returning_customers,
        returning_customers
    ]
})

customer_retention
```

```python
fig = px.bar(
    customer_retention,
    x="customer_type",
    y="customer_count",
    text="customer_count",
    title="Successful Customers by Customer Type"
)

fig.update_layout(
    xaxis_title="Customer Type",
    yaxis_title="Number of Customers"
)

fig.show()
```

### What does this chart mean?

Out of **20 successful customers**, **17 purchased only once** and only **3 returned to make another successful purchase**.

This gives a returning-customer rate of **15%**, which means the business currently depends mostly on one-time customers.

From a business perspective, the company should focus on customer-retention strategies such as loyalty rewards, follow-up offers, personalized recommendations, and promotions for previous customers. Increasing repeat purchases could improve long-term revenue without depending only on acquiring new customers.
